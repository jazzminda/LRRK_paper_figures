# Script to visualize data


Pull-down samples of TurboID, LRRK1-TurboID and LRRK2-TurboID samples combined with phosphoproteomics in HEK293T cells

Visualize two sample t-test results done in Perseus, data deposited in PRIDE repository under PXD067057 (Token: oBMf19A5vZpk; Username: reviewer_pxd067057@ebi.ac.uk; Password: 4ZCpKPoygWDs).


```python
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
import plotly.io as pio
import numpy as np
```

## Data input

```python
Int = pd.read_table('Int_inclTtests_forPython_revision_m741.txt', header=[0], skiprows=[1,2])

pInt = pd.read_table('pInt_inclTtests_forPython_m105.txt', header=[0], skiprows=[1,2])

pLys = pd.read_table('pLys_inclTtests_forPython_m291.txt', header=[0], skiprows=[1,2])
```

```python
fInt = Int.loc[Int["PG.NrOfPrecursorsIdentified (Experiment-wide)"] >1]
```

```python
# Int.shape, fInt.shape, pInt.shape, pLys.shape
```

```python
for df in (pInt, pLys):
    for gene in ("LRRK1", "LRRK2"):
        df[f"{gene} site"] = ""
        df.loc[df["Gene"].str.contains(gene, na=False), f"{gene} site"] = gene
```

## Figure S8

Combine the pInt and Int dataframes by using Accessions to create phospho vs protein fold change plot

```python
Int['Gene'] = Int['PG.Genes']
Int['Accession'] = Int['PG.ProteinAccessions']

merge = pd.merge(pInt, Int, on= ["Gene"])

fmerge_ML = merge.loc[merge["-Log Student\'s T-test p-value L2T_M_L2T_x"] > 1.3]
fmerge_RN = merge.loc[merge["-Log Student\'s T-test p-value L2T_R_L2T_x"] > 1.3]
```
